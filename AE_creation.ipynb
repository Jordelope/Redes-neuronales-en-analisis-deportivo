## Creación de autoencoders

En este notebook se da un guión para crear y guardar modelos de la clase Autoencoder, compuestos por dos redes MLP (encoder y decoder) definidas en este proyecto.
Se puede utilizar tanto para crear nuevos modelos desde cero como para ensamblar un autoencoder a partir de MLPs ya existentes.


Funcionalidad principal:
- Configuración de la arquitectura del encoder y decoder (número de entradas, espacio latente, capas ocultas, funciones de activación).
- Opción de cargar MLPs previamente guardados o crear nuevos modelos.
- Instanciación del autoencoder a partir de los MLPs definidos.
- Guardado del autoencoder y, opcionalmente, de los MLPs por separado.
- Posibilidad de añadir una descripción personalizada al modelo.


Uso:
1. Ajustar los parámetros de arquitectura, nombres de archivos y opciones de guardado según tus necesidades.
2. Ejecutar las celdas del notebook para crear y guardar el modelo Autoencoder.
3. El modelo podrá ser entrenado y utilizado posteriormente con otros scripts del proyecto.

### Preparación

Empezamos por importar las librerias, clases y funciones que vamos a usar.

```python
import torch
import torch.nn.functional as F
import random
from MLP import MLP
from Autoencoder import Autoencoder
from Guardar_Cargar import guardar_modelo, cargar_modelo
```

### Customización del modelo

Elegir nombre del archivo y ruta para el modelo y añadir descripción si lo desea.

```python
archivo_autoencoder = r"redes_disponibles\pruebas\prueba.json"        # Nombre del archivo(recomendado .json)
añadir_descripcion = True        # Opción de añadir una descripción
descripcion = "input_sz = 80, lat_spc_dim = 2 "      
```

Seleccionar si se van a importar MLPs ya creados para usar como encoder y decoder.

```python
existen_MLP = False       # En caso de True se importaran el encoder y decoder de archivos ya existentes.
archivo_encod = r"" 
archivo_decod = r"" 
```

#### Opciones guardado

Decidir si quiere guardar solo el autoencoder, también los submodelos, o no guardar ninguno y simplemente hacer pruebas.

```python
save_autoencoder = True         # En caso de True guarda el Autoencoder
save_decoder = False            # En caso de True guarda el decoder en un fichero por separado
save_encoder = False            # En caso de True guarda el encoder en un fichero por separado
```

#### Arquitectura del modelo

En esta sección establecemos la estructura de capas que tendrá el modelo. Eligiendo el numero de neuronas por capa y funciones de activación tanto del encoder como el autoencoder.

    Nota: Si se van a importar el encoder y decoder, esta parte es irrelevante. Pero es importante asegurarse que los MLPs coincidan en sus dimensiones.







Lo primero es determinar el tamaño de las entradas que recibirá el autoencoder y el tamaño del espacio latente del modelo.

```python
input_sz = 80            # Número de entradas
lat_spc_dim = 2         # Dimension espacio latente(salida encoder, entrada decoder) 
```

A continuacíon determinamos el número de neuronas por capa del encoder y del decoder. (Es habitual pero no necesario que la estructura de capas del decoder sea la inversa a la del encoder)

```python
estructura_encod = [ 64, 16]                   # Capas ocultas encoder
estructura_decod = estructura_encod[::-1]      # Capas ocultas decoder 
```

Ahora seleccionamos las funciones de activacíon del modelo especificandolas para cada uno de los submodelos.
Recuerde que la lista de funciones de activacion debe tener longitud = len(estructura_oct) + 1.

```python
# Funciones activacion encoder (None = [None,...,None] por defecto lineal en MLP)
lista_act_encod = [F.elu , F.leaky_relu] +[None] 

# Funciones activacion encoder (None = [None,...,None] por defecto lineal en MLP)
lista_act_decod = [ F.leaky_relu, F.elu] + [None] 
```

### Instanciación del modelo

Procedemos a definir el autoencoder según los atributos y arquitecturas diseñadas.

El primer paso es definir el encoder y decoder según si los estamos imprtando o definiendo la arquitectura.

```python
if existen_MLP:
        # Cargar MLP existentes
        encoder = cargar_modelo(archivo_encod)
        decoder = cargar_modelo(archivo_decod)
else:
        # Crear nuevos MLP para encoder y decoder
        encoder = MLP(input_sz, lat_spc_dim, estructura_encod, lista_act_encod)
        decoder = MLP(lat_spc_dim, input_sz, estructura_decod, lista_act_decod)
    
```

El siguiente paso es crear nuestro modelo

```python
autoencoder = Autoencoder(encoder, decoder)
# Añadimos descripción si así lo hemos decidido
if añadir_descripcion:
    autoencoder.description = descripcion
```

### Guardado

Una vez creado el modelo solo queda guardarlo en la ruta seleccionada y ya está listo para entrenar y usar.

```python
if save_autoencoder:
        guardar_modelo(autoencoder, archivo_autoencoder)
        if save_encoder :
            guardar_modelo(encoder, archivo_encod)
        if save_decoder:
            guardar_modelo(decoder, archivo_decod) 
else:
        print(f"Se ha decidido no guardar la red {archivo_autoencoder}.\n")
```
